Portfolio notebook connecting classical classification, ensembles, clustering and anomaly detection — the four principal workflows recovered from the source notebooks.


```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from src.classical_models import logistic_pipeline, decision_tree
from src.ensembles import random_forest_classifier
from src.evaluation import classification_summary, ks_statistic
from src.clustering import kmeans_scan
from src.anomaly_detection import isolation_forest_scores
```

```python
X, y = make_classification(n_samples=3500, n_features=10, n_informative=6, weights=[0.84, 0.16], class_sep=1.1, random_state=42)
X = pd.DataFrame(X, columns=[f'x{i}' for i in range(X.shape[1])])
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.3, stratify=y, random_state=42)
models = {'logistic': logistic_pipeline(), 'tree': decision_tree(), 'forest': random_forest_classifier()}
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    p = model.predict_proba(X_test)[:, 1]
    results[name] = {**classification_summary(y_test, p), 'ks': ks_statistic(y_test, p)}
pd.DataFrame(results).T
```
